### **DATA VISUALIZATION** 

##### **“How has the unemployment rate changed over time?”**

```python
sort_dates=emp_data.sort_values(by='Date')

plt.figure(figsize=(10,6))
sns.lineplot(
    data=sort_dates,
    x="Date",
    y="Estimated Unemployment Rate (%)"
)


plt.title('Unemployment Rate Over time')
plt.xlabel('Date')
plt.ylabel('Unemployment Rate (%)')
plt.show()
```

##### **“How did unemployment change before, during, and after Covid-19?"**


```python
sort_dates=emp_data.sort_values(by='Date')

covid_start=pd.to_datetime('2020-03-01')
covid_end=pd.to_datetime('2020-06-30')

plt.figure(figsize=(10,6))
sns.lineplot(data=sort_dates,x='Date',y='Estimated Unemployment Rate (%)')
plt.axvspan(covid_start, covid_end, color="red", alpha=0.2, label="Covid Period")
plt.legend()
plt.title("Unemployment Rate Before, During, and After Covid-19")
plt.xlabel("Date")
plt.ylabel("Unemployment Rate (%)")
plt.show()
```

##### **“Are there any seasonal patterns in unemployment across months?"**


```python
emp_data['Month']=emp_data['Date'].dt.month

plt.figure(figsize=(10,6))
sns.boxplot(data=emp_data,
           x="Month",y='Estimated Unemployment Rate (%)')
plt.title("Seasonal Pattern of Unemployment (Month-wise)")
plt.xlabel("Month")
plt.ylabel("Unemployment Rate (%)")
plt.show()
```

#### **"Which months consistently show higher unemployment rates?"**

```python
monthly_avg = (
    emp_data.groupby("Month")["Estimated Unemployment Rate (%)"]
    .mean()
    .sort_values(ascending=False)
)

plt.figure(figsize=(10,4))
bars = plt.bar(monthly_avg.index, monthly_avg.values)


for bar in bars:
    plt.text(
        bar.get_x() + bar.get_width() / 2,
        bar.get_height(),
        f'{bar.get_height():.2f}',
        ha='center', va='bottom', fontsize=8
    )
plt.xticks(monthly_avg.index)

plt.title("Average Unemployment Rate by Month (Sorted)")
plt.xlabel("Month")
plt.ylabel("Average Unemployment Rate (%)")
plt.show()
```

#### **"How does the unemployment rate differ across regions?"**

```python
plt.figure(figsize=(10,6))
sns.boxplot(data=emp_data,
           x='Region',
           y="Estimated Unemployment Rate (%)")
plt.title("Unemployment Rate Across Regions")
plt.xlabel("Region")
plt.ylabel("Unemployment Rate (%)")
plt.xticks(rotation=-90)
plt.show()
```

#### **"Which regions experienced the highest unemployment during Covid-19?"**

```python
covid_start= pd.to_datetime("2020-03-01")
covid_end=pd.to_datetime("2020-06-30")

covid_df=emp_data[(emp_data['Date']>=covid_start) & (emp_data['Date']<=covid_end)]
region_covid_avg=covid_df.groupby("Region")["Estimated Unemployment Rate (%)"].mean().sort_values(ascending=False).reset_index()

plt.figure(figsize=(10,6))
sns.barplot(data=region_covid_avg,
           x='Region',
           y="Estimated Unemployment Rate (%)")
plt.xticks(rotation=-90)
plt.title("Average Unemployment Rate by Region During Covid-19")
plt.xlabel("Region")
plt.ylabel("Unemployment Rate (%)")
plt.show()
```

INSIGHTS:
1.Certain regions show significantly higher unemployment during Covid-19 than others.
2.Regions with higher averages were more severely impacted by lockdowns.
3.Variations across regions indicate uneven economic impact of Covid-19.
4.Some regions remained relatively less affected, showing better resilience.
5.The results highlight the need for targeted, region-specific recovery policies.

#### **"Did all regions recover at the same pace after Covid-19?"**

```python
post_covid_start=pd.to_datetime("2020-07-01")
post_df=emp_data[emp_data['Date']>=post_covid_start]

plt.figure(figsize=(10,6))
sns.lineplot(data=post_df,
            x="Date",y="Estimated Unemployment Rate (%)",hue="Region")
plt.title("Post-Covid Unemployment Recovery by Region")
plt.xlabel("Date")
plt.ylabel("Unemployment Rate (%)")
plt.show()
```

```python
post_covid_start = "2020-07-01"
post_df = emp_data[emp_data["Date"] >= post_covid_start]

fig = px.line(
    post_df,
    x="Date",
    y="Estimated Unemployment Rate (%)",
    color="Region",
    title="Post-Covid Unemployment Recovery by Region"
)

fig.show()
```

INSIGHTS:
1.Post-Covid recovery was uneven across regions, with noticeable differences in recovery speed.
2.Some regions show a faster and steadier decline in unemployment after mid-2020.
3.Other regions remain volatile, indicating prolonged economic stress.
4.Interactive visualization reveals that no single recovery pattern fits all regions.
5.These differences highlight the need for region-specific post-Covid employment policies.

#### **"How volatile is the unemployment rate over time?"**

```python
date_sort=emp_data.sort_values('Date').copy()
date_sort['Rolling_std']=emp_data["Estimated Unemployment Rate (%)"].rolling(window=3).std()

plt.figure(figsize=(10,4))
sns.lineplot(data=date_sort, x="Date", y="Rolling_std")
plt.title("Unemployment Volatility Over Time (Rolling Std)")
plt.xlabel("Date")
plt.ylabel("Volatility (Std Dev)")
plt.show()
```

1.Unemployment volatility was low before 2020, indicating stable employment conditions.
2.A sharp spike in volatility appears during early 2020, reflecting sudden job losses due to Covid-19.
3.The highest volatility occurs around April–May 2020, showing extreme instability during lockdowns.
4.Volatility declines gradually after mid-2020, suggesting easing economic shocks.
5.Overall, unemployment experienced temporary but severe instability, followed by a slow return to stability.

#### **"What were the highest and lowest unemployment rates observed in the dataset?"**

```python
max_val = emp_data["Estimated Unemployment Rate (%)"].max()
min_val = emp_data["Estimated Unemployment Rate (%)"].min()

max_row = emp_data.loc[emp_data["Estimated Unemployment Rate (%)"].idxmax()]
min_row = emp_data.loc[emp_data["Estimated Unemployment Rate (%)"].idxmin()]

max_val, min_val, max_row[["Date","Region"]], min_row[["Date","Region"]]
```

INSIGHTS:
1.The highest unemployment rate observed is very high, indicating an extreme shock period.
2.This peak occurs during early 2020, aligning with Covid-19 lockdowns.
3.The lowest unemployment rate is comparatively small, reflecting normal/stable periods.
4.The wide gap between max and min confirms strong variability in unemployment.
5.Extreme values highlight how external shocks can dramatically affect employment.

#### **"What insights from unemployment trends can inform economic or social policies?"**

1.Sharp unemployment spikes during Covid show the need for emergency job-protection policies during crises.
2.Slow post-Covid recovery suggests that employment does not bounce back automatically, highlighting the importance of long-term recovery programs.
3.High regional differences indicate that region-specific employment policies are more effective than one uniform policy.
4.High volatility periods show the vulnerability of informal and temporary jobs, emphasizing the need for job security and social safety nets.
5.Seasonal and monthly patterns suggest that governments can plan targeted employment schemes during historically weak periods.
